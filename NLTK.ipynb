```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

```python
corpus = ["deep learning and machine learning","machine learning is deep"]
# 2. Initialize CountVectorizer (Leave binary=False to count occurrences)

bow_vectorizer = CountVectorizer()

```

```python
# 3. Fit and transform the corpus into frequencies
bow_matrix = bow_vectorizer.fit_transform(corpus)
```

```python
# 4. Extract unique words (Our vocabulary)
vocab = bow_vectorizer.get_feature_names_out()
```

```python
# 5. Represent it as a clean table
df_bow = pd.DataFrame(bow_matrix.toarray(), columns=vocab)
```

```python
print("Vocabulary List:", list(vocab))
print("\nBag of Words Matrix:")
print(df_bow)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Define a small corpus (notice how the word "learning" repeats)
corpus = [
    "deep learning is great",
    "machine learning is complex"
]

# 2. Initialize TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()

# 3. Fit and transform the corpus into TF-IDF scores
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)

# 4. Extract unique words (Our vocabulary features)
vocab = tfidf_vectorizer.get_feature_names_out()

# 5. Represent it as a clean table, rounding numbers to 2 decimals
df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=vocab)

print("Vocabulary List:\n", list(vocab))
print("\nTF-IDF Weighted Matrix:")
print(df_tfidf.round(2))
```
